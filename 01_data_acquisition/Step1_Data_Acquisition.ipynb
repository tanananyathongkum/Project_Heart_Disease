```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = pd.read_csv('heart_disease.csv')
display(data.head())
```

```python
print('--- Amount of Data and Feature ---')
display(data.shape)
```

```python
print('--- Information Type Details ---')
data.info()
```

```python
print('--- Statistical Summary ---')
display(data.describe())
```

```python
print('--- Missing Values in Each Feature ---')
display(data.isnull().sum())
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(data['Age'], kde=True, color='skyblue')
plt.title('--- Age Distribution ---')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 2)
sns.boxplot(x='Heart Disease Status', y='Cholesterol Level', data=data, palette='Set2')
plt.title('Cholesterol Level vs Heart Disease')
plt.tight_layout()
plt.show()
```

```python
cat_columns = [
    'Gender',
    'Smoking',
    'Diabetes',
    'Family Heart Disease',
    'Exercise Habits',
    'Stress Level'
]
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(15, 18))
axes = axes.flatten()
for i, col in enumerate(cat_columns):
    if i < len(axes):
        # สร้างตารางไขว้ (Crosstab) เพื่อนับจำนวน
        ct = pd.crosstab(data[col], data['Heart Disease Status'], normalize='index') 
        # วาดกราฟ Stacked Bar
        ct.plot(kind='bar', stacked=True, ax=axes[i], color=['#66c2a5', '#fc8d62'])

        # ตกแต่งกราฟ
        axes[i].set_title(f'Heart Disease by {col}', fontsize=14)
        axes[i].set_ylabel('Proportion (สัดส่วน)')
        axes[i].set_xlabel('')
        axes[i].legend(title='Heart Disease', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
