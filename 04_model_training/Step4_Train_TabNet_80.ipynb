```python
import pandas as pd
import numpy as np
from pytorch_tabnet.tab_model import TabNetClassifier
import torch
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, roc_curve, precision_score, recall_score, f1_score
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.pipeline import Pipeline
import seaborn as sns
import matplotlib.pyplot as plt
import joblib
import os
```

```python
import pandas as pd

path = '/Users/tanananyathongkum/Jenny/Project/Heart Disease/PJ 2/Data_Split/' 

# โหลดไฟล์
X_train = pd.read_csv(f'{path}X_train_80.csv')
X_test = pd.read_csv(f'{path}X_test_20.csv')
y_train = pd.read_csv(f'{path}y_train_80.csv').squeeze()
y_test = pd.read_csv(f'{path}y_test_20.csv').squeeze()

print("✅ โหลดชุดข้อมูล 80:20 เรียบร้อย!")
print(f"Train shape: {X_train.shape}")
print(f"Test shape: {X_test.shape}")
```

```python
print("-" * 30)
print('Baseline Model: TabNet')
print("-" * 30)

# Set random seeds
torch.manual_seed(42)
np.random.seed(42)

# Initialize TabNet model (Base parameters)
tabnet_params = dict(
    n_d=64, n_a=64, n_steps=5, gamma=1.5,
    n_independent=2, n_shared=2, lambda_sparse=1e-4,
    optimizer_fn=torch.optim.Adam,
    optimizer_params=dict(lr=2e-2),
    mask_type='entmax',
    scheduler_params=dict(mode="min", patience=5, min_lr=1e-5, factor=0.5),
    scheduler_fn=torch.optim.lr_scheduler.ReduceLROnPlateau,
    seed=42,
    verbose=1
)

# Cross-validation (store best fold model instead of retraining on full set)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = []

# track best fold
best_model_tabnet_base = None
best_score_tabnet_base = -1.0
best_fold_idx_tabnet_base = -1

print("\n" + "-" * 30)
print("Performing Cross-Validation...")
print("-" * 30)

for fold, (train_idx, val_idx) in enumerate(cv.split(X_train, y_train.values.ravel())):
    X_train_fold = X_train.iloc[train_idx].values
    X_val_fold = X_train.iloc[val_idx].values
    
    y_train_fold = y_train.values.ravel()[train_idx]
    y_val_fold = y_train.values.ravel()[val_idx]
    
    # Create model for this fold
    model_fold = TabNetClassifier(**tabnet_params)
    model_fold.verbose = 0
    
    # Train
    model_fold.fit(
        X_train_fold, y_train_fold,
        eval_set=[(X_val_fold, y_val_fold)],
        eval_metric=['auc'],
        max_epochs=100,
        patience=10,
        batch_size=256,
        virtual_batch_size=128
    )
    
    # Evaluate
    y_pred_fold = model_fold.predict(X_val_fold)
    score = accuracy_score(y_val_fold, y_pred_fold)
    cv_scores.append(score)
    print(f"Fold {fold+1}: {score:.4f}  ({score*100:.2f}%)")
    
    # keep best model from CV
    if score > best_score_tabnet_base:
        best_score_tabnet_base = score
        best_model_tabnet_base = model_fold
        best_fold_idx_tabnet_base = fold + 1
        # save best fold model
        os.makedirs('tabnet/models', exist_ok=True)
        best_model_tabnet_base.save_model(f'tabnet/models/TabNet_baseline_fold{best_fold_idx_tabnet_base}')

print("-" * 30)
print(f'>> Average CV Accuracy: {np.mean(cv_scores):.4f} (± {np.std(cv_scores):.4f})')
print("-" * 30)

# Use the best fold model from CV for test evaluation (do not retrain on full set)
if best_model_tabnet_base is None:
    raise RuntimeError('No best model found from CV. Check CV loop.')

print("\n" + "-" * 30)
print(f"Selected model from fold {best_fold_idx_tabnet_base} for test evaluation (saved: tabnet/models/TabNet_baseline_fold{best_fold_idx_tabnet_base}.zip)")
print("-" * 30)

# Ensure test arrays
X_test_array = X_test.values
y_test_array = y_test.values.ravel()

# Predictions with best fold model
y_pred = best_model_tabnet_base.predict(X_test_array)
y_pred_proba = best_model_tabnet_base.predict_proba(X_test_array)[:, 1]

# Test set performance
test_accuracy = accuracy_score(y_test_array, y_pred)
test_precision = precision_score(y_test_array, y_pred)
test_recall = recall_score(y_test_array, y_pred)
test_f1 = f1_score(y_test_array, y_pred)
test_roc_auc = roc_auc_score(y_test_array, y_pred_proba)

print("\n" + "=" * 50)
print("TEST SET PERFORMANCE")
print("=" * 50)
print(f"Accuracy:  {test_accuracy:.4f}")
print(f"Precision: {test_precision:.4f}")
print(f"Recall:    {test_recall:.4f}")
print(f"F1-Score:  {test_f1:.4f}")
print(f"ROC-AUC:   {test_roc_auc:.4f}")
print("=" * 50)

print("\nClassification Report:")
print(classification_report(y_test_array, y_pred, target_names=['No Disease', 'Disease']))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test_array, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='cividis', 
            xticklabels=['No Disease', 'Disease'], 
            yticklabels=['No Disease', 'Disease'],
            cbar_kws={'label': 'Count'})
plt.xlabel('Predicted', fontsize=12)
plt.ylabel('Actual', fontsize=12)
plt.title(f'Confusion Matrix - TabNet (Baseline) Fold {best_fold_idx_tabnet_base}', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Summary dictionary
tabnet_baseline_results = {
    'Model': 'TabNet',
    'Sampling': 'None (Baseline)',
    'CV_Accuracy': np.mean(cv_scores),
    'CV_Std': np.std(cv_scores),
    'Test_Accuracy': test_accuracy,
    'Precision': test_precision,
    'Recall': test_recall,
    'F1_Score': test_f1,
    'ROC_AUC': test_roc_auc
}

print("\n" + "=" * 50)
print("BASELINE RESULTS SUMMARY")
print("=" * 50)
for key, value in tabnet_baseline_results.items():
    if isinstance(value, float):
        print(f"{key:20s}: {value:.4f}")
    else:
        print(f"{key:20s}: {value}")
print("=" * 50)

print(f"✓ Saved: tabnet/models/TabNet_baseline_fold{best_fold_idx_tabnet_base}.zip")
```

```python
from pytorch_tabnet.tab_model import TabNetClassifier
from imblearn.over_sampling import SMOTE


print("-" * 30)
print('TabNet + SMOTE')
print("-" * 30)

# --- 1. เตรียมข้อมูลเป็น Numpy Array ---
X_train_np = X_train.values if hasattr(X_train, 'values') else X_train
y_train_np = y_train.values.ravel() if hasattr(y_train, 'values') else y_train
X_test_np = X_test.values if hasattr(X_test, 'values') else X_test
y_test_np = y_test.values.ravel() if hasattr(y_test, 'values') else y_test

# --- 2. ตั้งค่า Parameters ---
tabnet_params = dict(
    n_d=64, n_a=64, n_steps=5, gamma=1.5,
    n_independent=2, n_shared=2, lambda_sparse=1e-4,
    optimizer_fn=torch.optim.Adam,
    optimizer_params=dict(lr=2e-2),
    mask_type='entmax',
    scheduler_params=dict(mode="min", patience=5, min_lr=1e-5, factor=0.5),
    scheduler_fn=torch.optim.lr_scheduler.ReduceLROnPlateau,
    seed=42,
    verbose=0
)

# --- 3. SMOTE sampler (we'll still save this) ---
smote = SMOTE(random_state=42)

# (optional) show dataset sizes after applying SMOTE to full training set
X_train_smote, y_train_smote = smote.fit_resample(X_train_np, y_train_np)
print(f"Original training set size: {len(X_train_np)}")
print(f"After SMOTE training set size: {len(X_train_smote)}")
print(f"Original class distribution: {dict(pd.Series(y_train_np).value_counts())}")
print(f"After SMOTE class distribution: {dict(pd.Series(y_train_smote).value_counts())}")
print("-" * 30)

# --- 4. Cross-Validation with SMOTE; keep best fold model for testing ---
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = []

best_model_tabnet_smote = None
best_score_tabnet_smote = -1.0
best_fold_idx_tabnet_smote = -1

print("\n" + "-" * 30)
print("Performing Cross-Validation with SMOTE...")
print("-" * 30)

for fold, (train_idx, val_idx) in enumerate(cv.split(X_train_np, y_train_np)):
    X_train_fold = X_train_np[train_idx]
    y_train_fold = y_train_np[train_idx]
    X_val_fold = X_train_np[val_idx]
    y_val_fold = y_train_np[val_idx]
    
    # Apply SMOTE only on training fold
    X_train_fold_smote, y_train_fold_smote = smote.fit_resample(X_train_fold, y_train_fold)
    
    # Create and train model for this fold
    model_fold = TabNetClassifier(**tabnet_params)
    model_fold.fit(
        X_train_fold_smote, y_train_fold_smote,
        eval_set=[(X_val_fold, y_val_fold)],
        eval_metric=['auc'],
        max_epochs=100,
        patience=10,
        batch_size=256,
        virtual_batch_size=128
    )
    
    # Evaluate
    y_pred_fold = model_fold.predict(X_val_fold)
    score = accuracy_score(y_val_fold, y_pred_fold)
    cv_scores.append(score)
    print(f"Fold {fold+1}: {score:.4f}")

    # keep best fold model
    if score > best_score_tabnet_smote:
        best_score_tabnet_smote = score
        best_model_tabnet_smote = model_fold
        best_fold_idx_tabnet_smote = fold + 1
        os.makedirs('tabnet/models', exist_ok=True)
        best_model_tabnet_smote.save_model(f'tabnet/models/TabNet_smote_fold{best_fold_idx_tabnet_smote}')

print("-" * 30)
print(f'>> Average CV Accuracy: {np.mean(cv_scores):.4f} (± {np.std(cv_scores):.4f})')
print("-" * 30)

# --- 5. Use best CV fold model for test evaluation (do NOT retrain on full set) ---
if best_model_tabnet_smote is None:
    raise RuntimeError('No best model found from CV. Check CV loop for SMOTE.')

print("\n" + "-" * 30)
print(f"Selected model from fold {best_fold_idx_tabnet_smote} for test evaluation (saved: tabnet/models/TabNet_smote_fold{best_fold_idx_tabnet_smote}.zip)")
print("-" * 30)

# Predictions
y_pred = best_model_tabnet_smote.predict(X_test_np)
y_pred_proba = best_model_tabnet_smote.predict_proba(X_test_np)[:, 1]

# Test set performance
test_accuracy = accuracy_score(y_test_np, y_pred)
test_precision = precision_score(y_test_np, y_pred)
test_recall = recall_score(y_test_np, y_pred)
test_f1 = f1_score(y_test_np, y_pred)
test_roc_auc = roc_auc_score(y_test_np, y_pred_proba)

print("\n" + "=" * 50)
print("TEST SET PERFORMANCE (SMOTE)")
print("=" * 50)
print(f"Accuracy:  {test_accuracy:.4f}")
print(f"Precision: {test_precision:.4f}")
print(f"Recall:    {test_recall:.4f}")
print(f"F1-Score:  {test_f1:.4f}")
print(f"ROC-AUC:   {test_roc_auc:.4f}")
print("=" * 50)

print("\nClassification Report:")
print(classification_report(y_test_np, y_pred, target_names=['No Disease', 'Disease']))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test_np, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Greens', 
            xticklabels=['No Disease', 'Disease'], 
            yticklabels=['No Disease', 'Disease'],
            cbar_kws={'label': 'Count'})
plt.xlabel('Predicted', fontsize=12)
plt.ylabel('Actual', fontsize=12)
plt.title(f'Confusion Matrix - TabNet + SMOTE (Fold {best_fold_idx_tabnet_smote})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Summary dictionary
tabnet_smote_results = {
    'Model': 'TabNet',
    'Sampling': 'SMOTE',
    'CV_Accuracy': np.mean(cv_scores),
    'CV_Std': np.std(cv_scores),
    'Test_Accuracy': test_accuracy,
    'Precision': test_precision,
    'Recall': test_recall,
    'F1_Score': test_f1,
    'ROC_AUC': test_roc_auc
}

print("\n" + "=" * 50)
print("SMOTE RESULTS SUMMARY")
print("=" * 50)
for key, value in tabnet_smote_results.items():
    if isinstance(value, float):
        print(f"{key:20s}: {value:.4f}")
    else:
        print(f"{key:20s}: {value}")
print("=" * 50)

# Save sampler and confirm saved best-fold model
os.makedirs('tabnet/models', exist_ok=True)
joblib.dump(smote, 'tabnet/models/smote_sampler_tabnet.pkl')
print(f"✓ Saved: tabnet/models/TabNet_smote_fold{best_fold_idx_tabnet_smote}.zip")
print("✓ Saved: tabnet/models/smote_sampler_tabnet.pkl")
```

```python
from imblearn.over_sampling import BorderlineSMOTE

print("-" * 30)
print('TabNet + BorderlineSMOTE + Tomek Links')
print("-" * 30)

# --- 1. Borderline+Tomek pipeline (will be saved) ---
pipe_bt = ImbPipeline([
    ('borderline', BorderlineSMOTE(random_state=42, kind='borderline-2')),
    ('tomek', TomekLinks())
])

# (optional) show overall resampled sizes for information
try:
    print("Resampling with Borderline+Tomek Links for info...")
    X_train_bt_check, y_train_bt_check = pipe_bt.fit_resample(X_train_np, y_train_np)
    print(f"Original size: {len(X_train_np)}")
    print(f"After Borderline+Tomek size: {len(X_train_bt_check)}")
    print("-" * 30)
except Exception as e:
    print(f"Warning: could not compute full-set resample summary: {e}")
    print("Continuing with CV-resampling only.")

# --- 2. Cross-Validation; keep best fold model ---
cv_scores = []
best_model_tabnet_bt = None
best_score_tabnet_bt = -1.0
best_fold_idx_tabnet_bt = -1

print("\nPerforming Cross-Validation with Borderline+Tomek...")

for fold, (train_idx, val_idx) in enumerate(cv.split(X_train_np, y_train_np)):
    X_train_fold, y_train_fold = X_train_np[train_idx], y_train_np[train_idx]
    X_val_fold, y_val_fold = X_train_np[val_idx], y_train_np[val_idx]
    
    # Resample training fold
    try:
        X_train_fold_bt, y_train_fold_bt = pipe_bt.fit_resample(X_train_fold, y_train_fold)
    except ValueError as e:
        print(f"Warning in fold {fold+1} during resampling: {e}")
        continue
    
    # Train
    model_fold = TabNetClassifier(**tabnet_params)
    model_fold.fit(
        X_train_fold_bt, y_train_fold_bt,
        eval_set=[(X_val_fold, y_val_fold)],
        eval_metric=['auc'],
        max_epochs=100, patience=10, batch_size=256, virtual_batch_size=128
    )
    score = accuracy_score(y_val_fold, model_fold.predict(X_val_fold))
    cv_scores.append(score)
    print(f"Fold {fold+1}: {score:.4f}")

    # keep best fold model
    if score > best_score_tabnet_bt:
        best_score_tabnet_bt = score
        best_model_tabnet_bt = model_fold
        best_fold_idx_tabnet_bt = fold + 1
        os.makedirs('tabnet/models', exist_ok=True)
        best_model_tabnet_bt.save_model(f'tabnet/models/TabNet_borderline_tomek_fold{best_fold_idx_tabnet_bt}')

print("-" * 30)
if len(cv_scores) > 0:
    print(f'>> Average CV Accuracy: {np.mean(cv_scores):.4f} (± {np.std(cv_scores):.4f})')
else:
    print('>> No CV scores (all folds skipped during resampling).')
print("-" * 30)

# --- 3. Use best CV fold model for test evaluation (do NOT retrain on full set) ---
if best_model_tabnet_bt is None:
    raise RuntimeError('No best model found from CV. Check CV loop for Borderline+Tomek.')

print("\n" + "-" * 30)
print(f"Selected model from fold {best_fold_idx_tabnet_bt} for test evaluation (saved: tabnet/models/TabNet_borderline_tomek_fold{best_fold_idx_tabnet_bt}.zip)")
print("-" * 30)

# --- 4. Evaluation ---
y_pred = best_model_tabnet_bt.predict(X_test_np)
y_pred_proba = best_model_tabnet_bt.predict_proba(X_test_np)[:, 1]

test_accuracy = accuracy_score(y_test_np, y_pred)
test_precision = precision_score(y_test_np, y_pred)
test_recall = recall_score(y_test_np, y_pred)
test_f1 = f1_score(y_test_np, y_pred)
test_roc_auc = roc_auc_score(y_test_np, y_pred_proba)

# --- Display Results ---
print("\n" + "=" * 50)
print("TEST SET PERFORMANCE (Borderline + Tomek)")
print("=" * 50)
print(f"Accuracy:  {test_accuracy:.4f}")
print(f"Precision: {test_precision:.4f}")
print(f"Recall:    {test_recall:.4f}")
print(f"F1-Score:  {test_f1:.4f}")
print(f"ROC-AUC:   {test_roc_auc:.4f}")
print("=" * 50)

print("\nClassification Report:")
print(classification_report(y_test_np, y_pred, target_names=['No Disease', 'Disease']))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test_np, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Oranges', 
            xticklabels=['No Disease', 'Disease'], 
            yticklabels=['No Disease', 'Disease'],
            cbar_kws={'label': 'Count'})
plt.xlabel('Predicted', fontsize=12)
plt.ylabel('Actual', fontsize=12)
plt.title(f'Confusion Matrix - TabNet + Borderline + Tomek (Fold {best_fold_idx_tabnet_bt})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Summary dictionary
tabnet_borderline_tomek_results = {
    'Model': 'TabNet', 'Sampling': 'Borderline+Tomek',
    'CV_Accuracy': np.mean(cv_scores) if len(cv_scores) > 0 else float('nan'), 
    'CV_Std': np.std(cv_scores) if len(cv_scores) > 0 else float('nan'),
    'Test_Accuracy': test_accuracy, 'Precision': test_precision,
    'Recall': test_recall, 'F1_Score': test_f1, 'ROC_AUC': test_roc_auc
}

print("\n" + "=" * 50)
print("BORDERLINE + TOMEK RESULTS SUMMARY")
print("=" * 50)
for key, value in tabnet_borderline_tomek_results.items():
    if isinstance(value, float):
        print(f"{key:20s}: {value:.4f}")
    else:
        print(f"{key:20s}: {value}")
print("=" * 50)

# Save sampler and confirm saved best-fold model
os.makedirs('tabnet/models', exist_ok=True)
joblib.dump(pipe_bt, 'tabnet/models/borderline_tomek_sampler_tabnet.pkl')
print(f"✓ Saved: tabnet/models/TabNet_borderline_tomek_fold{best_fold_idx_tabnet_bt}.zip")
print("✓ Saved: tabnet/models/borderline_tomek_sampler_tabnet.pkl")
```
